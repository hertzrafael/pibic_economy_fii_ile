```python
import pandas as pd

economic_freedom = pd.read_excel(
    './files/index_economic_freedom_frasier_2000-2022.xlsx',
    skiprows=4
)
economic_freedom
```

```python
economic_freedom.columns
```

```python
economic_freedom = (economic_freedom.drop(economic_freedom.columns[[0]], axis=1)
    .dropna(subset=['Year'])
    .astype({'Year': int})
)
economic_freedom.head(10)
```

```python
(economic_freedom
    .filter(items=['Year', 'Countries', ' Economic Freedom Summary Index', 'Rank'])
    .query('Countries == "Brazil"')
)
```

```python
esg_data = pd.read_excel('files/dados_empresas_esg.xls')
esg_data.head(5)
```

```python
min_year = esg_data['Ano'].min()
max_year = esg_data['Ano'].max()

print(min_year, '-', max_year)
```

```python
freedom_filtered = (economic_freedom
        .filter(items=['ISO Code 3', 'Year', 'Countries', ' Economic Freedom Summary Index', 'Rank'])
        .query(f"Year >= {min_year} and Year <= {max_year}")
)
print(freedom_filtered['Year'].min(), freedom_filtered['Year'].max())
```

## Dados descritivos ESG de países desenvolvidos e emergentes

```python
enterprises_esg_refinitiv_describe = esg_data.groupby('ID')['ESGRefinitiv'].describe().reset_index()

target_columns = [
    'País', 'Name', 'ID', 'G'
]

esg_target_group = esg_data.filter(items=target_columns).drop_duplicates(subset=['ID'])

described_enterprises = pd.merge(left=enterprises_esg_refinitiv_describe, right=esg_target_group, how='left', on='ID')
described_enterprises
```

```python
(described_enterprises
 .assign(Emergente=described_enterprises['G'].apply(lambda x: 'Sim' if x == 0 else 'Não'))
 .drop(columns=['G'])
 .rename(columns={
     'ID': 'id',
     'País': 'country',
     'Name': 'name',
     'Emergente': 'emergent'
 })
 .to_excel('results/empresas_dados_descritivos_refinitiv.xlsx', index=False)
)
```

## Describe do ESGRefinitiv dos Países

```python
country_esg_refinitiv_describe = esg_data.groupby('País')['ESGRefinitiv'].describe().reset_index()
country_esg_refinitiv_describe
```

```python
country_esg_refinitiv_describe.to_excel('results/paises_dados_descritivos_refinitiv.xlsx', index=False)
```

## Juntando dados sobre ESG e Liberdade Econômica em um DataFrame

```python
countries_iso_code_three = pd.read_csv('files/countries_iso_code_3.csv')
countries_iso_code_three
```

```python
esg_refinitiv_iso_countries = pd.merge(left=countries_iso_code_three, right=country_esg_refinitiv_describe, how='left', on='País')
esg_refinitiv_iso_countries
```

```python
mean_ile = freedom_filtered.groupby('ISO Code 3')[' Economic Freedom Summary Index'].mean().reset_index()
mean_ile
```

```python
esg_ile_countries = (pd
        .merge(left=esg_refinitiv_iso_countries, right=mean_ile, left_on='ISO3', right_on='ISO Code 3', how='left')
        .filter(items=['País', 'ISO3', 'mean', ' Economic Freedom Summary Index'])
        .rename(columns={
            ' Economic Freedom Summary Index': 'economic_freedom_index',
            'País': 'country',
            'ISO3': 'iso3',
            'mean': 'mean_esg_refinitiv'
        })
)

esg_ile_countries
```

```python
esg_ile_countries.to_excel('results/paises_esg_ief.xlsx', index=False)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

plt.scatter(esg_ile_countries['economic_freedom_index'], esg_ile_countries['mean_esg_refinitiv'], color='blue')

plt.xlabel('Liberdade Econômica')
plt.ylabel('Score ESG')
plt.title('Liberdade Econômica vs ESG (Países)')

plt.grid(True)
plt.show()
```
